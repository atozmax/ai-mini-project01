# 3. Problem Scenario

Credit card fraud is a major challenge for financial institutions.

A bank or payment service provider wants to develop a machine learning system that can identify potentially fraudulent credit card transactions.

Your task is to build an end-to-end machine learning pipeline that predicts whether a transaction is fraudulent.

The model should classify each transaction into one of two classes:

```text

0 → Legitimate Transaction

1 → Fraudulent Transaction

```
The main challenge is that fraudulent transactions are much less frequent than legitimate transactions.

Therefore, achieving high Accuracy alone is not sufficient.

Your goal is to build a model that can reliably detect fraudulent transactions while carefully analyzing the trade-off between:

- False Positives
- False Negatives
- Precision
- Recall
- F1-score

---


# 4. Dataset Description

## Credit Card Fraud Detection Dataset

Source:

https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud

The dataset contains credit card transactions made by European cardholders.

The goal is to identify fraudulent transactions using machine learning.

### Dataset Characteristics

| Property | Value |
|---|---|
| Number of Samples | 284,807 |
| Number of Features | 30 Features |
| Target | Class |
| Problem Type | Binary Classification |
| Fraudulent Transactions | 492 |
| Legitimate Transactions | 284,315 |
| Fraud Ratio | ~0.17% |

### Target Variable

The `Class` column is the target variable:

```text
0 → Legitimate Transaction
1 → Fraudulent Transaction
```

The dataset contains:

- Time
- V1 to V28
- Amount

**Features**

The features V1 to V28 are anonymized numerical features resulting from a transformation of the original transaction data.

Because these features are anonymized, the focus of this project is on:

- Data analysis
- Classification
- Class imbalance
- Model evaluation
- Experimentation
- Reliable fraud detection

# Phase 2: Data Preprocessing

You must correctly perform:

- Data quality checks
- Missing value analysis
- Duplicate analysis
- Train/Test Split
- Feature Scaling

Because this is an imbalanced classification problem, use a stratified train/test split.

For example:

```python
train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

## Important Rule: Avoid Data Leakage

The following approach is incorrect:

```

Scaling
|
Train/Test Split

```


Because information from the test set is used during preprocessing.


The correct approach:

```

Train/Test Split
|
v
Fit Scaler on Training Data
|
v
Transform Train and Test Data

